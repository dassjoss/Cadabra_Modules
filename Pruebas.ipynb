```python
def expand_cov_deriv(ex):
    # Contador para generar índices dummy (p1, p2...) y no repetir
    idx_counter = [0]
    
    for nabla in ex[r'\nabla']:
        # 1. \nabla se vuelve una derivada parcial plana \partial
        nabla.name = r'\partial'
        dindex = nabla.indices().__next__()
        
        # 2. Recorremos los argumentos
        for arg in nabla.args():
            ret := 0;
            # 3. Recorremos TODOS los índices libres de este tensor
            for index in arg.free_indices():
                t2 := @(arg);
                idx_counter[0] += 1
                newidx = Ex(r'p' + str(idx_counter[0]))
                
                # Si el índice está abajo (subíndice) -> restamos conexión
                if index.parent_rel == sub:
                    t1 := -\Gamma^{@(newidx)}_{@(dindex) @(index)};
                    t2[index] := _{@(newidx)};
                # Si el índice está arriba (superíndice) -> sumamos conexión
                else:
                    t1 := \Gamma^{@(index)}_{@(dindex) @(newidx)};
                    t2[index] := ^{@(newidx)};
                
                # Sumamos el término al total
                ret += Ex(str(nabla.multiplier)) * t1 * t2
                
            # Agregamos los términos de conexión a la derivada parcial
            nabla += ret
    return ex
```

## Codigo de la IA con modificaicon de problema


```python
import cadabra2
from cadabra2 import Ex, parent_rel_t
from cdb.utils.indices import get_free_indices


# =============================================================================
# UTILIDADES
# =============================================================================

def get_full_index_name(n):
    """
    Reconstruye el nombre completo de un índice.

    Ejemplo:
        parser:
            name  = \\lambda
            child = 1

        devuelve:
            \\lambda_{1}
    """

    name = n.name

    children = list(n.children())

    if children:
        name += "_{" + children[0].name + "}"

    return name


def normalizar_nombre_indice(nombre):
    """
    Hace equivalentes:

        \\lambda_1
        \\lambda_{1}

    usando siempre la forma:

        \\lambda_{1}
    """

    if "_" not in nombre:
        return nombre

    base, subindice = nombre.split("_", 1)

    subindice = subindice.strip()

    if (
        subindice.startswith("{")
        and
        subindice.endswith("}")
    ):
        subindice = subindice[1:-1]

    return base + "_{" + subindice + "}"


def obtener_nodo_indice(expr, nombre_buscado):

    nombre_buscado = normalizar_nombre_indice(
        nombre_buscado
    )

    iterador = expr.ex() if hasattr(expr, "ex") else expr

    for n in iterador:

        rel = str(
            getattr(
                n,
                "parent_rel",
                ""
            )
        )

        if ("sub" in rel or "super" in rel):

            nombre_actual = normalizar_nombre_indice(
                get_full_index_name(n)
            )

            if nombre_actual == nombre_buscado:

                return n

    return None


def mutar_indice(expr, nombre_original, nombre_nuevo):

    nombre_original = normalizar_nombre_indice(
        nombre_original
    )

    iterador = expr.ex() if hasattr(expr, "ex") else expr

    for n in iterador:

        rel = str(
            getattr(
                n,
                "parent_rel",
                ""
            )
        )

        if ("sub" in rel or "super" in rel):

            nombre_actual = normalizar_nombre_indice(
                get_full_index_name(n)
            )

            if nombre_actual == nombre_original:

                n.name = nombre_nuevo

                for c in list(n.children()):
                    c.erase()


# =============================================================================
# d_c_g
# =============================================================================

def d_c_g(
    ex,
    idx_curvos,
    idx_planos,
    curvos_todos,
    planos_todos
):

    # =========================================================================
    # NORMALIZAR LOS ALFABETOS DE ÍNDICES
    # =========================================================================

    curvos_todos_norm = {
        normalizar_nombre_indice(idx): idx
        for idx in curvos_todos
    }

    planos_todos_norm = {
        normalizar_nombre_indice(idx): idx
        for idx in planos_todos
    }


    # =========================================================================
    # COPIAS DE TRABAJO
    # =========================================================================

    curvos_trabajo = idx_curvos.copy()
    planos_trabajo = idx_planos.copy()


    # =========================================================================
    # BUSCAR TODAS LAS DERIVADAS COVARIANTES
    # =========================================================================

    while True:

        iterador = ex[r'\nabla']

        try:

            nabla = next(iterador)

        except StopIteration:

            break


        # =====================================================================
        # ESCANEO GLOBAL DE ÍNDICES USADOS
        # =====================================================================

        indices_en_uso = set()

        iterador_global = (
            ex.ex()
            if hasattr(ex, "ex")
            else ex
        )

        for n in iterador_global:

            nombre = get_full_index_name(n)

            if nombre:

                indices_en_uso.add(
                    normalizar_nombre_indice(nombre)
                )


        # =====================================================================
        # REFRESCAR ÍNDICES AUXILIARES DISPONIBLES
        # =====================================================================

        curvos_trabajo = [
            idx
            for idx in curvos_todos
            if normalizar_nombre_indice(idx)
            not in indices_en_uso
        ]

        planos_trabajo = [
            idx
            for idx in planos_todos
            if normalizar_nombre_indice(idx)
            not in indices_en_uso
        ]


        # =====================================================================
        # ARGUMENTOS DE ∇
        # =====================================================================

        argumentos = list(
            nabla.args()
        )


        # =====================================================================
        # ∇ SIN ARGUMENTOS → ∂
        # =====================================================================

        if len(argumentos) == 0:

            nabla.name = r'\partial'

            continue


        # =====================================================================
        # ÍNDICE DE DERIVACIÓN
        # =====================================================================

        dindex = nabla.indices().__next__()


        # =====================================================================
        # TÉRMINOS DE CONEXIÓN
        # =====================================================================

        terminos_conexion = Ex(0)


        # =====================================================================
        # PROCESAR CADA ARGUMENTO DE ∇
        # =====================================================================

        for arg in argumentos:

            nodo_ex = Ex(
                r'@(arg)'
            )


            # =================================================================
            # OBTENER ÍNDICES LIBRES DEL ARGUMENTO
            # =================================================================

            try:

                indices_libres_cpp = get_free_indices(
                    nodo_ex
                )

                indices_logicos = [
                    get_full_index_name(
                        next(iter(idx))
                    )
                    for idx in indices_libres_cpp
                ]

            except Exception:

                continue


            # =================================================================
            # PROCESAR CADA ÍNDICE LIBRE
            # =================================================================

            for nombre_idx in indices_logicos:

                idx_original = obtener_nodo_indice(
                    arg,
                    nombre_idx
                )

                if not idx_original:

                    continue


                # =============================================================
                # NORMALIZAR NOMBRE PARA LAS COMPARACIONES
                # =============================================================

                nombre_idx_norm = normalizar_nombre_indice(
                    nombre_idx
                )


                # =============================================================
                # ¿ES UN ÍNDICE CURVO?
                # =============================================================

                if nombre_idx_norm not in curvos_todos_norm:

                    continue


                # =============================================================
                # ¿QUEDA UN ÍNDICE AUXILIAR DISPONIBLE?
                # =============================================================

                if len(curvos_trabajo) == 0:

                    raise RuntimeError(
                        "d_c_g: no quedan índices auxiliares curvos "
                        "disponibles para generar el término de conexión."
                    )


                # =============================================================
                # ELEGIR ÍNDICE MUDO
                # =============================================================

                mudo = curvos_trabajo.pop(0)


                # =============================================================
                # COPIA DEL ARGUMENTO
                # =============================================================

                t2 = Ex(
                    r'@(arg)'
                )


                # =============================================================
                # CAMBIAR:
                #
                #     A^{lambda}
                #
                # POR:
                #
                #     A^{lambda_1}
                #
                # TEMPORALMENTE
                # =============================================================

                mutar_indice(
                    t2,
                    nombre_idx,
                    mudo
                )


                # =============================================================
                # REPARAR EL ÁRBOL SINTÁCTICO
                #
                # IMPORTANTE:
                #
                # mutar_indice() puede producir:
                #
                #     name = \\lambda_1
                #
                # en un solo nodo.
                #
                # Al volver a pasar por Ex(str(...)), Cadabra reconstruye:
                #
                #     name = \\lambda
                #     child = 1
                #
                # =============================================================

                t2 = Ex(
                    str(t2)
                )


                # =============================================================
                # CONSTRUIR EL GAMMA
                #
                # SI EL ÍNDICE DEL TENSOR ES SUPERIOR:
                #
                #     Γ^{alpha}_{mu lambda}
                #
                # SI ES INFERIOR:
                #
                #     -Γ^{lambda}_{mu alpha}
                # =============================================================

                if idx_original.parent_rel == parent_rel_t.sub:

                    t1 = Ex(
                        r"-\Gamma^{" +
                        mudo +
                        r"}_{" +
                        dindex.name +
                        " " +
                        nombre_idx +
                        r"}"
                    )

                else:

                    t1 = Ex(
                        r"\Gamma^{" +
                        nombre_idx +
                        r"}_{" +
                        dindex.name +
                        " " +
                        mudo +
                        r"}"
                    )


                # =============================================================
                # AGREGAR EL TÉRMINO DE CONEXIÓN
                # =============================================================

                terminos_conexion = (
                    terminos_conexion
                    +
                    t1 * t2
                )


        # =====================================================================
        # ∇ → ∂
        # =====================================================================

        nabla.name = r'\partial'


        # =====================================================================
        # CONSTRUIR:
        #
        #     ∂ + términos de conexión
        # =====================================================================

        reemplazo = (
            Ex(
                r'@(nabla)'
            )
            +
            terminos_conexion
        )


        # =====================================================================
        # REEMPLAZAR EL NODO ∇
        # =====================================================================

        nabla.replace(
            reemplazo
        )


    return None
```

```python
from cadabra2 import Ex
import re


# ============================================================
# EXTRAER ÍNDICES DESDE TEXTO
# ============================================================

def extraer_indices(texto):
    """
    Extrae los índices directamente desde la representación
    textual de una expresión Cadabra.

    Retorna:

        (representacion, nombre_base, posicion)

    Ejemplos:

        A^{\mu}
            -> [('μ', 'mu', 'super')]

        B_{\rho}
            -> [('ρ', 'rho', 'sub')]

        C^{\lambda_{1}}
            -> [('λ_{1}', 'lambda', 'super')]

        A^{\mu}_{\rho}
            -> [
                ('μ', 'mu', 'super'),
                ('ρ', 'rho', 'sub')
            ]
    """

    indices = []

    traduccion = {
        r'\mu': 'μ',
        r'\nu': 'ν',
        r'\rho': 'ρ',
        r'\sigma': 'σ',
        r'\lambda': 'λ',
    }

    # ========================================================
    # FUNCIÓN AUXILIAR
    # ========================================================

    def procesar_tokens(contenido, posicion):

        tokens = re.findall(
            r'\\[a-zA-Z]+(?:_\{[^{}]+\})?'
            r'|\\[a-zA-Z]+'
            r'|[A-Za-zΑ-Ωα-ωμνρσλ]+',
            contenido
        )

        resultado = []

        for token in tokens:

            # ------------------------------------------------
            # Nombre base
            # ------------------------------------------------

            if token.startswith('\\'):

                base = token.split("_{", 1)[0]

                nombre = base[1:]

                representacion = token

                # ------------------------------------------------
                # Convertir comandos LaTeX conocidos
                # ------------------------------------------------

                if base in traduccion:

                    simbolo = traduccion[base]

                    if "_{" in token:

                        subindice = token[
                            token.find("_{") + 2:-1
                        ]

                        representacion = (
                            f"{simbolo}_{{{subindice}}}"
                        )

                    else:

                        representacion = simbolo

            else:

                nombre = token
                representacion = token

            resultado.append(
                (
                    representacion,
                    nombre,
                    posicion
                )
            )

        return resultado

    # ========================================================
    # ÍNDICES SUPERIORES
    # ========================================================

    contenidos_superiores = re.findall(
        r'\^\{([^{}]*(?:\{[^{}]*\}[^{}]*)*)\}',
        texto
    )

    for contenido in contenidos_superiores:

        indices.extend(
            procesar_tokens(
                contenido,
                "super"
            )
        )

    # ========================================================
    # ÍNDICES INFERIORES
    # ========================================================

    contenidos_inferiores = re.findall(
        r'_\{([^{}]*(?:\{[^{}]*\}[^{}]*)*)\}',
        texto
    )

    for contenido in contenidos_inferiores:

        indices.extend(
            procesar_tokens(
                contenido,
                "sub"
            )
        )

    return indices


# ============================================================
# OBTENER ÍNDICES DE UNA EXPRESIÓN INDIVIDUAL
# ============================================================

def obtener_indices_original(expr):

    # --------------------------------------------------------
    # IMPORTANTE:
    #
    # No usamos .indices().
    #
    # Convertimos la expresión a texto y reconstruimos una
    # expresión independiente. De esta manera evitamos depender
    # de los cursores internos de Cadabra.
    # --------------------------------------------------------

    texto = str(expr)

    return extraer_indices(texto)


# ============================================================
# OBTENER ÍNDICES LIBRES
# ============================================================

def obtener_indices_libres(expr):

    # --------------------------------------------------------
    # COPIA INDEPENDIENTE DEL TOP
    # --------------------------------------------------------

    nodo_original = expr.top()

    copia = Ex(
        str(nodo_original)
    )

    nodo = copia.top()

    indices_totales = []

    print(
        "TOP:",
        nodo
    )

    print(
        "NAME:",
        nodo.name
    )

    # ========================================================
    # CASO 1: PRODUCTO
    # ========================================================

    if nodo.name == r'\prod':

        for argumento in nodo.args():

            argumento_copia = Ex(
                str(argumento)
            ).top()

            indices = obtener_indices_original(
                argumento_copia
            )

            print(
                "\n" + "=" * 50
            )

            print(
                "FACTOR:",
                argumento_copia
            )

            print(
                "ÍNDICES:",
                indices
            )

            indices_totales.extend(
                indices
            )

    # ========================================================
    # CASO 2: NODO INDIVIDUAL
    # ========================================================

    else:

        indices_totales = obtener_indices_original(
            nodo
        )

        print(
            "\n" + "=" * 50
        )

        print(
            "NODO:",
            nodo
        )

        print(
            "ÍNDICES:",
            indices_totales
        )

    # ========================================================
    # RESULTADO ACTUAL
    # ========================================================

    print(
        "\n" + "=" * 50
    )

    print(
        "ÍNDICES TOTALES:"
    )

    print(
        indices_totales
    )

    return indices_totales
```
