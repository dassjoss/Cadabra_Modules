```cadabra2
def expand_cov_deriv(ex):
    # Contador para generar índices dummy (p1, p2...) y no repetir
    idx_counter = [0]
    
    for nabla in ex[r'\nabla']:
        # 1. \nabla se vuelve una derivada parcial plana \partial
        nabla.name = r'\partial'
        dindex = nabla.indices().__next__()
        
        # 2. Recorremos los argumentos
        for arg in nabla.args():
            ret := 0;
            # 3. Recorremos TODOS los índices libres de este tensor
            for index in arg.free_indices():
                t2 := @(arg);
                idx_counter[0] += 1
                newidx = Ex(r'p' + str(idx_counter[0]))
                
                # Si el índice está abajo (subíndice) -> restamos conexión
                if index.parent_rel == sub:
                    t1 := -\Gamma^{@(newidx)}_{@(dindex) @(index)};
                    t2[index] := _{@(newidx)};
                # Si el índice está arriba (superíndice) -> sumamos conexión
                else:
                    t1 := \Gamma^{@(index)}_{@(dindex) @(newidx)};
                    t2[index] := ^{@(newidx)};
                
                # Sumamos el término al total
                ret += Ex(str(nabla.multiplier)) * t1 * t2
                
            # Agregamos los términos de conexión a la derivada parcial
            nabla += ret
    return ex
```

## Codigo de la IA con modificaicon de problema


```cadabra2
import cadabra2
from cadabra2 import Ex, parent_rel_t
from cdb.utils.indices import get_free_indices


# =============================================================================
# UTILIDADES
# =============================================================================

def get_full_index_name(n):
    """
    Reconstruye el nombre completo de un índice.

    Ejemplo:
        parser:
            name  = \\lambda
            child = 1

        devuelve:
            \\lambda_{1}
    """

    name = n.name

    children = list(n.children())

    if children:
        name += "_{" + children[0].name + "}"

    return name


def normalizar_nombre_indice(nombre):
    """
    Hace equivalentes:

        \\lambda_1
        \\lambda_{1}

    usando siempre la forma:

        \\lambda_{1}
    """

    if "_" not in nombre:
        return nombre

    base, subindice = nombre.split("_", 1)

    subindice = subindice.strip()

    if (
        subindice.startswith("{")
        and
        subindice.endswith("}")
    ):
        subindice = subindice[1:-1]

    return base + "_{" + subindice + "}"


def obtener_nodo_indice(expr, nombre_buscado):

    nombre_buscado = normalizar_nombre_indice(
        nombre_buscado
    )

    iterador = expr.ex() if hasattr(expr, "ex") else expr

    for n in iterador:

        rel = str(
            getattr(
                n,
                "parent_rel",
                ""
            )
        )

        if ("sub" in rel or "super" in rel):

            nombre_actual = normalizar_nombre_indice(
                get_full_index_name(n)
            )

            if nombre_actual == nombre_buscado:

                return n

    return None


def mutar_indice(expr, nombre_original, nombre_nuevo):

    nombre_original = normalizar_nombre_indice(
        nombre_original
    )

    iterador = expr.ex() if hasattr(expr, "ex") else expr

    for n in iterador:

        rel = str(
            getattr(
                n,
                "parent_rel",
                ""
            )
        )

        if ("sub" in rel or "super" in rel):

            nombre_actual = normalizar_nombre_indice(
                get_full_index_name(n)
            )

            if nombre_actual == nombre_original:

                n.name = nombre_nuevo

                for c in list(n.children()):
                    c.erase()


# =============================================================================
# d_c_g
# =============================================================================

def d_c_g(
    ex,
    idx_curvos,
    idx_planos,
    curvos_todos,
    planos_todos
):

    # =========================================================================
    # NORMALIZAR LOS ALFABETOS DE ÍNDICES
    # =========================================================================

    curvos_todos_norm = {
        normalizar_nombre_indice(idx): idx
        for idx in curvos_todos
    }

    planos_todos_norm = {
        normalizar_nombre_indice(idx): idx
        for idx in planos_todos
    }


    # =========================================================================
    # COPIAS DE TRABAJO
    # =========================================================================

    curvos_trabajo = idx_curvos.copy()
    planos_trabajo = idx_planos.copy()


    # =========================================================================
    # BUSCAR TODAS LAS DERIVADAS COVARIANTES
    # =========================================================================

    while True:

        iterador = ex[r'\nabla']

        try:

            nabla = next(iterador)

        except StopIteration:

            break


        # =====================================================================
        # ESCANEO GLOBAL DE ÍNDICES USADOS
        # =====================================================================

        indices_en_uso = set()

        iterador_global = (
            ex.ex()
            if hasattr(ex, "ex")
            else ex
        )

        for n in iterador_global:

            nombre = get_full_index_name(n)

            if nombre:

                indices_en_uso.add(
                    normalizar_nombre_indice(nombre)
                )


        # =====================================================================
        # REFRESCAR ÍNDICES AUXILIARES DISPONIBLES
        # =====================================================================

        curvos_trabajo = [
            idx
            for idx in curvos_todos
            if normalizar_nombre_indice(idx)
            not in indices_en_uso
        ]

        planos_trabajo = [
            idx
            for idx in planos_todos
            if normalizar_nombre_indice(idx)
            not in indices_en_uso
        ]


        # =====================================================================
        # ARGUMENTOS DE ∇
        # =====================================================================

        argumentos = list(
            nabla.args()
        )


        # =====================================================================
        # ∇ SIN ARGUMENTOS → ∂
        # =====================================================================

        if len(argumentos) == 0:

            nabla.name = r'\partial'

            continue


        # =====================================================================
        # ÍNDICE DE DERIVACIÓN
        # =====================================================================

        dindex = nabla.indices().__next__()


        # =====================================================================
        # TÉRMINOS DE CONEXIÓN
        # =====================================================================

        terminos_conexion = Ex(0)


        # =====================================================================
        # PROCESAR CADA ARGUMENTO DE ∇
        # =====================================================================

        for arg in argumentos:

            nodo_ex = Ex(
                r'@(arg)'
            )


            # =================================================================
            # OBTENER ÍNDICES LIBRES DEL ARGUMENTO
            # =================================================================

            try:

                indices_libres_cpp = get_free_indices(
                    nodo_ex
                )

                indices_logicos = [
                    get_full_index_name(
                        next(iter(idx))
                    )
                    for idx in indices_libres_cpp
                ]

            except Exception:

                continue


            # =================================================================
            # PROCESAR CADA ÍNDICE LIBRE
            # =================================================================

            for nombre_idx in indices_logicos:

                idx_original = obtener_nodo_indice(
                    arg,
                    nombre_idx
                )

                if not idx_original:

                    continue


                # =============================================================
                # NORMALIZAR NOMBRE PARA LAS COMPARACIONES
                # =============================================================

                nombre_idx_norm = normalizar_nombre_indice(
                    nombre_idx
                )


                # =============================================================
                # ¿ES UN ÍNDICE CURVO?
                # =============================================================

                if nombre_idx_norm not in curvos_todos_norm:

                    continue


                # =============================================================
                # ¿QUEDA UN ÍNDICE AUXILIAR DISPONIBLE?
                # =============================================================

                if len(curvos_trabajo) == 0:

                    raise RuntimeError(
                        "d_c_g: no quedan índices auxiliares curvos "
                        "disponibles para generar el término de conexión."
                    )


                # =============================================================
                # ELEGIR ÍNDICE MUDO
                # =============================================================

                mudo = curvos_trabajo.pop(0)


                # =============================================================
                # COPIA DEL ARGUMENTO
                # =============================================================

                t2 = Ex(
                    r'@(arg)'
                )


                # =============================================================
                # CAMBIAR:
                #
                #     A^{lambda}
                #
                # POR:
                #
                #     A^{lambda_1}
                #
                # TEMPORALMENTE
                # =============================================================

                mutar_indice(
                    t2,
                    nombre_idx,
                    mudo
                )


                # =============================================================
                # REPARAR EL ÁRBOL SINTÁCTICO
                #
                # IMPORTANTE:
                #
                # mutar_indice() puede producir:
                #
                #     name = \\lambda_1
                #
                # en un solo nodo.
                #
                # Al volver a pasar por Ex(str(...)), Cadabra reconstruye:
                #
                #     name = \\lambda
                #     child = 1
                #
                # =============================================================

                t2 = Ex(
                    str(t2)
                )


                # =============================================================
                # CONSTRUIR EL GAMMA
                #
                # SI EL ÍNDICE DEL TENSOR ES SUPERIOR:
                #
                #     Γ^{alpha}_{mu lambda}
                #
                # SI ES INFERIOR:
                #
                #     -Γ^{lambda}_{mu alpha}
                # =============================================================

                if idx_original.parent_rel == parent_rel_t.sub:

                    t1 = Ex(
                        r"-\Gamma^{" +
                        mudo +
                        r"}_{" +
                        dindex.name +
                        " " +
                        nombre_idx +
                        r"}"
                    )

                else:

                    t1 = Ex(
                        r"\Gamma^{" +
                        nombre_idx +
                        r"}_{" +
                        dindex.name +
                        " " +
                        mudo +
                        r"}"
                    )


                # =============================================================
                # AGREGAR EL TÉRMINO DE CONEXIÓN
                # =============================================================

                terminos_conexion = (
                    terminos_conexion
                    +
                    t1 * t2
                )


        # =====================================================================
        # ∇ → ∂
        # =====================================================================

        nabla.name = r'\partial'


        # =====================================================================
        # CONSTRUIR:
        #
        #     ∂ + términos de conexión
        # =====================================================================

        reemplazo = (
            Ex(
                r'@(nabla)'
            )
            +
            terminos_conexion
        )


        # =====================================================================
        # REEMPLAZAR EL NODO ∇
        # =====================================================================

        nabla.replace(
            reemplazo
        )


    return None
```

```cadabra2
from cadabra2 import Ex

from scripts.obtener_indices_libres import obtener_indices_libres


# ============================================================
# UTILIDADES
# ============================================================

def ejecutar_test(nombre, expresion, esperado):
    print("=" * 100)
    print(nombre)
    print("=" * 100)

    print("EXPRESIÓN:")
    print(expresion)

    try:
        resultado = obtener_indices_libres(expresion)

        print("\nRESULTADO:")
        print(resultado)

        print("\nESPERADO:")
        print(esperado)

        if set(resultado) == set(esperado):
            print("\n✓ PASÓ")
            return True
        else:
            print("\n✗ FALLÓ")
            return False

    except Exception as e:
        print("\n✗ EXCEPCIÓN:")
        print(type(e).__name__, ":", e)
        return False


# ============================================================
# 1 — TENSOR SIMPLE
# ============================================================

A = Ex(r"A_{\mu\nu}")

ejecutar_test(
    "1 — TENSOR SIMPLE",
    A,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 2 — CONTRACCIÓN SIMPLE
# ============================================================

A = Ex(r"A^{\mu}_{\mu}")

ejecutar_test(
    "2 — CONTRACCIÓN SIMPLE",
    A,
    []
)


# ============================================================
# 3 — PRODUCTO
# ============================================================

A = Ex(r"A_{\mu\nu}")
B = Ex(r"B^{\nu}_{\rho}")

producto = A * B

ejecutar_test(
    "3 — PRODUCTO",
    producto,
    [
        ("μ", "sub"),
        ("ρ", "sub")
    ]
)


# ============================================================
# 4 — PRODUCTO CON CONTRACCIÓN
# ============================================================

A = Ex(r"A^{\mu}_{\nu}")
B = Ex(r"B^{\nu}_{\rho}")

producto = A * B

ejecutar_test(
    "4 — PRODUCTO CON CONTRACCIÓN",
    producto,
    [
        ("μ", "super"),
        ("ρ", "sub")
    ]
)


# ============================================================
# 5 — SUMA VÁLIDA
# ============================================================
#
# Ambos términos tienen exactamente los mismos índices libres.
# Cadabra acepta directamente la suma.
# ============================================================

T1 = Ex(r"A_{\mu\nu}")
T2 = Ex(r"B_{\mu\nu}")

suma = T1 - T2

ejecutar_test(
    "5 — SUMA TENSOR - TENSOR",
    suma,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 6 — DERIVADA CONTRAÍDA INDIVIDUAL
# ============================================================
#
# No construimos una suma inválida.
# Primero verificamos cada término individualmente.
# ============================================================

T1 = Ex(r"\partial_{\rho}(A^{\rho}_{\mu\nu})")

ejecutar_test(
    "6 — DERIVADA CONTRAÍDA",
    T1,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 7 — SUMA DE DERIVADAS CON MISMO PATRÓN
# ============================================================
#
# Esta suma sí es válida para Cadabra.
# ============================================================

T1 = Ex(r"\partial_{\rho}(A^{\rho}_{\mu\nu})")
T2 = Ex(r"\partial_{\rho}(B^{\rho}_{\mu\nu})")

suma = T1 - T2

ejecutar_test(
    "7 — SUMA DE DERIVADAS CON MISMO PATRÓN",
    suma,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 8 — OPERADOR SOBRE TENSOR
# ============================================================

nabla_A = Ex(
    r"\nabla_{\lambda}(A_{\mu\nu})"
)

ejecutar_test(
    "8 — OPERADOR SOBRE TENSOR",
    nabla_A,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 9 — OPERADOR SOBRE PRODUCTO
# ============================================================

nabla_producto = Ex(
    r"\nabla_{\lambda}"
    r"(A^{\rho}_{\mu}B_{\rho\nu})"
)

ejecutar_test(
    "9 — OPERADOR SOBRE PRODUCTO",
    nabla_producto,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 10 — OPERADOR SOBRE SUMA VÁLIDA
# ============================================================

T1 = Ex(r"A_{\mu\nu}")
T2 = Ex(r"B_{\mu\nu}")

suma = T1 - T2

nabla_suma = Ex(
    r"\nabla_{\lambda}(A_{\mu\nu}-B_{\mu\nu})"
)

ejecutar_test(
    "10 — OPERADOR SOBRE SUMA SIMPLE",
    nabla_suma,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 11 — OPERADOR SOBRE SUMA DE DERIVADAS
# ============================================================
#
# IMPORTANTE:
#
# La suma interna:
#
#   ∂ρ(A^ρ_{μν}) - ∂ν(B^ρ_{μρ})
#
# es problemática para Cadabra.
#
# Por eso NO hacemos:
#
#   Ex(r"\nabla_\lambda(...suma...)")
#
# directamente.
#
# Construimos primero cada término como Ex independiente,
# hacemos la resta a nivel Python y luego colocamos el
# operador externo.
# ============================================================

T1 = Ex(
    r"\partial_{\rho}(A^{\rho}_{\mu\nu})"
)

T2 = Ex(
    r"\partial_{\nu}(B^{\rho}_{\mu\rho})"
)

# La suma problemática NO se reconstruye como Ex(T1 - T2).

suma_problematica = T1 - T2

# El operador externo se construye usando la representación
# de la suma ya existente.

nabla_suma = Ex(
    r"\nabla_{\lambda}("
    + str(suma_problematica)
    + r")"
)

ejecutar_test(
    "11 — OPERADOR SOBRE SUMA DE DERIVADAS",
    nabla_suma,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 12 — TÉRMINOS INDIVIDUALES DE RICCI
# ============================================================
#
# Primero verificamos cada término por separado.
# ============================================================

R1 = Ex(
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
)

R2 = Ex(
    r"\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
)

R3 = Ex(
    r"\Gamma^{\rho}_{\mu\nu}"
    r"\Gamma^{\sigma}_{\rho\sigma}"
)

R4 = Ex(
    r"\Gamma^{\rho}_{\mu\sigma}"
    r"\Gamma^{\sigma}_{\nu\rho}"
)


ejecutar_test(
    "12A — RICCI TÉRMINO 1",
    R1,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "12B — RICCI TÉRMINO 2",
    R2,
    [
        ("μ", "sub")
    ]
)

ejecutar_test(
    "12C — RICCI TÉRMINO 3",
    R3,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "12D — RICCI TÉRMINO 4",
    R4,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 13 — RICCI COMPLETO COMO SUMA PROBLEMÁTICA
# ============================================================
#
# Cadabra NO debe construir directamente esta suma:
#
# ∂ρ Γ^ρ_{μν}
# - ∂ν Γ^ρ_{μρ}
# + Γ^ρ_{μν} Γ^σ_{ρσ}
# - Γ^ρ_{μσ} Γ^σ_{νρ}
#
# porque el segundo término tiene una representación de índices
# libres que Cadabra puede rechazar durante la construcción.
#
# Por tanto NO ejecutamos:
#
#     Ricci = Ex(r"...")
#
# ============================================================

# Construimos cada término independientemente.

R1 = Ex(
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
)

R2 = Ex(
    r"\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
)

R3 = Ex(
    r"\Gamma^{\rho}_{\mu\nu}"
    r"\Gamma^{\sigma}_{\rho\sigma}"
)

R4 = Ex(
    r"\Gamma^{\rho}_{\mu\sigma}"
    r"\Gamma^{\sigma}_{\nu\rho}"
)

# No intentamos construir Ricci como un Ex completo.
#
# En cambio verificamos que cada término tenga los índices
# matemáticamente esperados antes de aplicar cualquier operador.

print("=" * 100)
print("13 — RICCI: VERIFICACIÓN DE TÉRMINOS")
print("=" * 100)

for nombre, termino, esperado in [
    (
        "R1",
        R1,
        [("μ", "sub"), ("ν", "sub")]
    ),
    (
        "R2",
        R2,
        [("μ", "sub")]
    ),
    (
        "R3",
        R3,
        [("μ", "sub"), ("ν", "sub")]
    ),
    (
        "R4",
        R4,
        [("μ", "sub"), ("ν", "sub")]
    )
]:

    ejecutar_test(
        nombre,
        termino,
        esperado
    )


# ============================================================
# 14 — OPERADOR SOBRE RICCI
# ============================================================
#
# Este es el test más importante.
#
# Construimos los cuatro términos independientemente y luego
# construimos el operador externo sobre su representación.
#
# ============================================================

R1 = Ex(
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
)

R2 = Ex(
    r"\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
)

R3 = Ex(
    r"\Gamma^{\rho}_{\mu\nu}"
    r"\Gamma^{\sigma}_{\rho\sigma}"
)

R4 = Ex(
    r"\Gamma^{\rho}_{\mu\sigma}"
    r"\Gamma^{\sigma}_{\nu\rho}"
)

# Construimos los términos mediante operaciones entre Ex.
#
# Esto evita crear una expresión inicial mediante Ex(str(...))
# que fuerce a Cadabra a validar toda la suma simultáneamente.

Ricci = R1 - R2 + R3 - R4

# Ahora aplicamos el operador externamente.

nabla_Ricci = Ex(
    r"\nabla_{\lambda}("
    + str(Ricci)
    + r")"
)

ejecutar_test(
    "14 — OPERADOR SOBRE RICCI",
    nabla_Ricci,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)
```

```cadabra2
import sys
import os

sys.path.insert(0, os.path.abspath("."))

from cadabra2 import Ex

from scripts.obtener_indices_libres import obtener_indices_libres


# ============================================================
# BANCO DE PRUEBAS GENÉRICO — ÍNDICES LIBRES
# ============================================================

total = 0
pasaron = 0
fallaron = 0
excepciones = 0


def ejecutar_test(nombre, expresion, esperado):
    global total, pasaron, fallaron, excepciones

    total += 1

    print("=" * 100)
    print(f"{total:03d} — {nombre}")
    print("=" * 100)

    print("EXPRESIÓN:")
    print(expresion)

    try:
        resultado = obtener_indices_libres(expresion)

        resultado_set = set(resultado)
        esperado_set = set(esperado)

        print("\nRESULTADO:")
        print(resultado)

        print("\nESPERADO:")
        print(esperado)

        if resultado_set == esperado_set:
            print("\n✓ PASÓ")
            pasaron += 1
            return True

        print("\n✗ FALLÓ")
        fallaron += 1
        return False

    except Exception as e:
        print("\n✗ EXCEPCIÓN:")
        print(type(e).__name__, ":", e)

        excepciones += 1
        return False


# ============================================================
# 1. HOJAS — TENSORES
# ============================================================

ejecutar_test(
    "Tensor escalar",
    Ex(r"A"),
    []
)

ejecutar_test(
    "Vector covariante",
    Ex(r"A_{\mu}"),
    [("μ", "sub")]
)

ejecutar_test(
    "Vector contravariante",
    Ex(r"A^{\mu}"),
    [("μ", "super")]
)

ejecutar_test(
    "Tensor covariante de rango 2",
    Ex(r"A_{\mu\nu}"),
    [("μ", "sub"), ("ν", "sub")]
)

ejecutar_test(
    "Tensor mixto",
    Ex(r"A^{\mu}_{\nu}"),
    [("μ", "super"), ("ν", "sub")]
)

ejecutar_test(
    "Tensor contravariante de rango 2",
    Ex(r"A^{\mu\nu}"),
    [("μ", "super"), ("ν", "super")]
)

ejecutar_test(
    "Tensor covariante de rango 3",
    Ex(r"A_{\mu\nu\rho}"),
    [
        ("μ", "sub"),
        ("ν", "sub"),
        ("ρ", "sub")
    ]
)


# ============================================================
# 2. CONTRACCIONES
# ============================================================

ejecutar_test(
    "Una contracción",
    Ex(r"A^{\mu}_{\mu}"),
    []
)

ejecutar_test(
    "Dos contracciones",
    Ex(r"A^{\mu\nu}_{\mu\nu}"),
    []
)

ejecutar_test(
    "Una contracción + índice libre",
    Ex(r"A^{\mu\nu}_{\mu\rho}"),
    [("ν", "super"), ("ρ", "sub")]
)

ejecutar_test(
    "Dos índices libres superiores",
    Ex(r"A^{\mu\nu}_{\mu\rho}"),
    [("ν", "super"), ("ρ", "sub")]
)

ejecutar_test(
    "Contracción cruzada",
    Ex(r"A^{\mu}_{\nu}"),
    [("μ", "super"), ("ν", "sub")]
)


# ============================================================
# 3. PRODUCTOS
# ============================================================

A = Ex(r"A_{\mu}")
B = Ex(r"B^{\mu}")

ejecutar_test(
    "Producto completamente contraído",
    A * B,
    []
)


A = Ex(r"A_{\mu}")
B = Ex(r"B^{\mu}_{\nu}")

ejecutar_test(
    "Producto con un índice libre",
    A * B,
    [("ν", "sub")]
)


A = Ex(r"A^{\mu}")
B = Ex(r"B_{\mu\nu}")

ejecutar_test(
    "Producto con un índice libre inferior",
    A * B,
    [("ν", "sub")]
)


A = Ex(r"A^{\mu}_{\nu}")
B = Ex(r"B^{\nu}_{\rho}")

ejecutar_test(
    "Cadena de tensores",
    A * B,
    [
        ("μ", "super"),
        ("ρ", "sub")
    ]
)


A = Ex(r"A_{\mu\nu}")
B = Ex(r"B^{\nu\rho}")

ejecutar_test(
    "Producto con contracción intermedia",
    A * B,
    [
        ("μ", "sub"),
        ("ρ", "super")
    ]
)


# ============================================================
# 4. PRODUCTOS MÁS GRANDES
# ============================================================

A = Ex(r"A^{\mu}_{\nu}")
B = Ex(r"B^{\nu}_{\rho}")
C = Ex(r"C^{\rho}_{\sigma}")

ejecutar_test(
    "Cadena de tres tensores",
    A * B * C,
    [
        ("μ", "super"),
        ("σ", "sub")
    ]
)


A = Ex(r"A_{\mu\nu}")
B = Ex(r"B^{\nu\rho}")
C = Ex(r"C_{\rho\sigma}")

ejecutar_test(
    "Cadena de tres tensores con dos contracciones",
    A * B * C,
    [
        ("μ", "sub"),
        ("σ", "sub")
    ]
)


A = Ex(r"A^{\mu\nu}")
B = Ex(r"B_{\nu\rho}")
C = Ex(r"C^{\rho}_{\sigma}")

ejecutar_test(
    "Cadena mixta de tres tensores",
    A * B * C,
    [
        ("μ", "super"),
        ("σ", "sub")
    ]
)


# ============================================================
# 5. PRODUCTOS COMPLETAMENTE CONTRAÍDOS
# ============================================================

A = Ex(r"A^{\mu}_{\nu}")
B = Ex(r"B^{\nu}_{\mu}")

ejecutar_test(
    "Producto completamente contraído con dos pares",
    A * B,
    []
)


A = Ex(r"A^{\mu\nu}")
B = Ex(r"B_{\nu\mu}")

ejecutar_test(
    "Contracción doble cruzada",
    A * B,
    []
)


# ============================================================
# 6. DERIVADAS
# ============================================================

ejecutar_test(
    "Derivada de escalar",
    Ex(r"\partial_{\mu}(A)"),
    [("μ", "sub")]
)

ejecutar_test(
    "Derivada de vector",
    Ex(r"\partial_{\mu}(A_{\nu})"),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Derivada de tensor",
    Ex(r"\partial_{\lambda}(A_{\mu\nu})"),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Derivada contraída",
    Ex(r"\partial_{\rho}(A^{\rho}_{\mu})"),
    [("μ", "sub")]
)

ejecutar_test(
    "Derivada con dos índices",
    Ex(r"\partial_{\rho}(A^{\rho\mu}_{\nu})"),
    [
        ("μ", "super"),
        ("ν", "sub")
    ]
)


# ============================================================
# 7. SEGUNDAS DERIVADAS
# ============================================================

ejecutar_test(
    "Segunda derivada",
    Ex(r"\partial_{\mu}(\partial_{\nu}(A))"),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Segunda derivada contraída",
    Ex(r"\partial^{\mu}(\partial_{\mu}(A))"),
    []
)

ejecutar_test(
    "Segunda derivada con índice libre",
    Ex(r"\partial_{\mu}(\partial_{\nu}(A^{\mu}))"),
    [
        ("ν", "sub")
    ]
)


# ============================================================
# 8. OPERADORES COVARIANTES
# ============================================================

ejecutar_test(
    "Nabla sobre escalar",
    Ex(r"\nabla_{\mu}(A)"),
    [("μ", "sub")]
)

ejecutar_test(
    "Nabla sobre vector",
    Ex(r"\nabla_{\mu}(A_{\nu})"),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Nabla sobre vector contravariante",
    Ex(r"\nabla_{\mu}(A^{\nu})"),
    [
        ("μ", "sub"),
        ("ν", "super")
    ]
)

ejecutar_test(
    "Nabla sobre tensor",
    Ex(r"\nabla_{\lambda}(A_{\mu\nu})"),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 9. OPERADORES SOBRE PRODUCTOS
# ============================================================

ejecutar_test(
    "Nabla sobre producto escalar",
    Ex(r"\nabla_{\mu}(A B)"),
    [("μ", "sub")]
)

ejecutar_test(
    "Nabla sobre producto vectorial",
    Ex(r"\nabla_{\lambda}(A^{\rho}B_{\rho})"),
    [("λ", "sub")]
)

ejecutar_test(
    "Nabla sobre producto tensorial",
    Ex(r"\nabla_{\lambda}(A^{\rho}_{\mu}B_{\rho\nu})"),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 10. OPERADORES ANIDADOS
# ============================================================

ejecutar_test(
    "Nabla de derivada",
    Ex(r"\nabla_{\lambda}(\partial_{\mu}(A))"),
    [
        ("λ", "sub"),
        ("μ", "sub")
    ]
)

ejecutar_test(
    "Derivada de nabla",
    Ex(r"\partial_{\lambda}(\nabla_{\mu}(A_{\nu}))"),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Nabla doble",
    Ex(r"\nabla_{\lambda}(\nabla_{\mu}(A_{\nu}))"),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 11. SUMAS VÁLIDAS
# ============================================================

ejecutar_test(
    "Suma de vectores",
    Ex(r"A_{\mu}+B_{\mu}"),
    [("μ", "sub")]
)

ejecutar_test(
    "Suma de tensores",
    Ex(r"A_{\mu\nu}-B_{\mu\nu}"),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Suma de tensores mixtos",
    Ex(r"A^{\mu}_{\nu}+B^{\mu}_{\nu}"),
    [
        ("μ", "super"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Suma de derivadas con mismo patrón",
    Ex(
        r"\partial_{\rho}(A^{\rho}_{\mu\nu})"
        r"-\partial_{\rho}(B^{\rho}_{\mu\nu})"
    ),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 12. SUMAS CON PRODUCTOS
# ============================================================

ejecutar_test(
    "Suma de productos",
    Ex(
        r"A_{\mu}B_{\nu}"
        r"+C_{\mu}D_{\nu}"
    ),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Suma de productos mixtos",
    Ex(
        r"A^{\mu}B_{\nu}"
        r"-C^{\mu}D_{\nu}"
    ),
    [
        ("μ", "super"),
        ("ν", "sub")
    ]
)


# ============================================================
# 13. SUMAS CON OPERADORES
# ============================================================

ejecutar_test(
    "Suma de derivadas",
    Ex(
        r"\partial_{\mu}(A_{\nu})"
        r"+\partial_{\mu}(B_{\nu})"
    ),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)

ejecutar_test(
    "Suma de nablas",
    Ex(
        r"\nabla_{\mu}(A_{\nu})"
        r"-\nabla_{\mu}(B_{\nu})"
    ),
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 14. OFUSCACIÓN — OPERADOR SOBRE SUMA
# ============================================================

ejecutar_test(
    "Nabla sobre suma",
    Ex(
        r"\nabla_{\lambda}("
        r"A_{\mu\nu}-B_{\mu\nu}"
        r")"
    ),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


ejecutar_test(
    "Derivada sobre suma",
    Ex(
        r"\partial_{\lambda}("
        r"A_{\mu\nu}+B_{\mu\nu}"
        r")"
    ),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


ejecutar_test(
    "Nabla sobre suma de derivadas",
    Ex(
        r"\nabla_{\lambda}("
        r"\partial_{\rho}(A^{\rho}_{\mu\nu})"
        r"-\partial_{\rho}(B^{\rho}_{\mu\nu})"
        r")"
    ),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 15. CONTRACCIONES DENTRO DE OPERADORES
# ============================================================

ejecutar_test(
    "Nabla sobre contracción",
    Ex(
        r"\nabla_{\lambda}(A^{\rho}_{\rho})"
    ),
    [("λ", "sub")]
)

ejecutar_test(
    "Nabla sobre producto contraído",
    Ex(
        r"\nabla_{\lambda}(A^{\rho}_{\mu}B^{\mu}_{\rho})"
    ),
    [("λ", "sub")]
)

ejecutar_test(
    "Nabla sobre doble contracción",
    Ex(
        r"\nabla_{\lambda}(A^{\mu\nu}_{\mu\nu})"
    ),
    [("λ", "sub")]
)


# ============================================================
# 16. PRODUCTOS DENTRO DE OPERADORES
# ============================================================

ejecutar_test(
    "Nabla sobre cadena",
    Ex(
        r"\nabla_{\lambda}("
        r"A^{\rho}_{\mu}"
        r"B^{\sigma}_{\rho}"
        r"C_{\sigma\nu}"
        r")"
    ),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


ejecutar_test(
    "Derivada sobre cadena",
    Ex(
        r"\partial_{\lambda}("
        r"A^{\rho}_{\mu}"
        r"B^{\sigma}_{\rho}"
        r"C_{\sigma\nu}"
        r")"
    ),
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 17. RICCI — TÉRMINOS INDIVIDUALES
# ============================================================

R1 = Ex(
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
)

ejecutar_test(
    "Ricci — R1",
    R1,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


R2 = Ex(
    r"\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
)

ejecutar_test(
    "Ricci — R2",
    R2,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


R3 = Ex(
    r"\Gamma^{\rho}_{\mu\nu}"
    r"\Gamma^{\sigma}_{\rho\sigma}"
)

ejecutar_test(
    "Ricci — R3",
    R3,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


R4 = Ex(
    r"\Gamma^{\rho}_{\mu\sigma}"
    r"\Gamma^{\sigma}_{\nu\rho}"
)

ejecutar_test(
    "Ricci — R4",
    R4,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 18. RICCI VÁLIDO
# ============================================================
#
# Se construye directamente mediante Ex porque todos sus
# términos tienen los mismos índices libres.
# ============================================================

Ricci = Ex(
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
    r"-\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
    r"+\Gamma^{\rho}_{\mu\nu}\Gamma^{\sigma}_{\rho\sigma}"
    r"-\Gamma^{\rho}_{\mu\sigma}\Gamma^{\sigma}_{\nu\rho}"
)

ejecutar_test(
    "Tensor de Ricci completo",
    Ricci,
    [
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 19. RICCI DENTRO DE OPERADOR
# ============================================================

nabla_Ricci = Ex(
    r"\nabla_{\lambda}("
    r"\partial_{\rho}(\Gamma^{\rho}_{\mu\nu})"
    r"-\partial_{\nu}(\Gamma^{\rho}_{\mu\rho})"
    r"+\Gamma^{\rho}_{\mu\nu}\Gamma^{\sigma}_{\rho\sigma}"
    r"-\Gamma^{\rho}_{\mu\sigma}\Gamma^{\sigma}_{\nu\rho}"
    r")"
)

ejecutar_test(
    "Nabla del tensor de Ricci",
    nabla_Ricci,
    [
        ("λ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 20. CASOS CON ÍNDICES LIBRES SUPERIORES
# ============================================================

ejecutar_test(
    "Operador sobre tensor con índices superiores",
    Ex(
        r"\nabla_{\lambda}(A^{\mu\nu})"
    ),
    [
        ("λ", "sub"),
        ("μ", "super"),
        ("ν", "super")
    ]
)


ejecutar_test(
    "Derivada sobre tensor mixto",
    Ex(
        r"\partial_{\lambda}(A^{\mu}_{\nu})"
    ),
    [
        ("λ", "sub"),
        ("μ", "super"),
        ("ν", "sub")
    ]
)


# ============================================================
# 21. CADENAS MÁS PROFUNDAS
# ============================================================

ejecutar_test(
    "Tres operadores anidados",
    Ex(
        r"\nabla_{\alpha}("
        r"\partial_{\beta}("
        r"\nabla_{\gamma}(A_{\mu\nu})"
        r"))"
    ),
    [
        ("α", "sub"),
        ("β", "sub"),
        ("γ", "sub"),
        ("μ", "sub"),
        ("ν", "sub")
    ]
)


ejecutar_test(
    "Operadores + producto + contracción",
    Ex(
        r"\nabla_{\lambda}("
        r"\partial_{\rho}("
        r"A^{\rho}_{\mu}B^{\mu}_{\nu}"
        r"))"
    ),
    [
        ("λ", "sub"),
        ("ν", "sub")
    ]
)


# ============================================================
# 22. EXPRESIONES TIPO EINSTEIN
# ============================================================

ejecutar_test(
    "Divergencia de vector",
    Ex(
        r"\nabla_{\mu}(A^{\mu})"
    ),
    []
)


ejecutar_test(
    "Divergencia de tensor",
    Ex(
        r"\nabla_{\mu}(A^{\mu}_{\nu})"
    ),
    [("ν", "sub")]
)


ejecutar_test(
    "Divergencia contravariante",
    Ex(
        r"\partial_{\mu}(A^{\mu})"
    ),
    []
)


ejecutar_test(
    "Laplaciano escalar",
    Ex(
        r"\nabla^{\mu}(\nabla_{\mu}(A))"
    ),
    []
)


# ============================================================
# RESUMEN
# ============================================================

print("\n")
print("=" * 100)
print("RESUMEN FINAL")
print("=" * 100)

print(f"TOTAL:       {total}")
print(f"PASARON:     {pasaron}")
print(f"FALLARON:    {fallaron}")
print(f"EXCEPCIONES: {excepciones}")

if fallaron == 0 and excepciones == 0:
    print("\n✓✓✓ TODOS LOS TESTS PASARON ✓✓✓")
else:
    print("\n✗ HAY TESTS QUE REQUIEREN REVISIÓN")
```
